## 1. My lane (or freestyle) and why

**Our Question:** Can we use the last 90 days of search numbers (like page views, positions, and user scroll rates) to rank our web pages and find the ones that are losing traffic the fastest?

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**What decision does your work improve?**
It helps us decide exactly which web pages our writing team should spend their limited time updating each week.

**Who acts on it?**
A Content Operations Manager or Editorial Team Lead will act on it by taking our top-ranked list and assigning those pages to writers.

**What does a wrong recommendation cost?**
* **If we flag a healthy page by mistake (False Positive):** We waste valuable budget and hours paying a writer to rewrite a page that didn't need any work.
* **If we miss a dying page by mistake (False Negative):** High-value pages quietly rot and lose their ranks, causing the company to lose search traffic and revenue to competitors.

```python
# Let's see the average value (Cost Per Click) of the pages we are monitoring (This shows what a wrong recommendation could cost)
average_cpc = df_starter['cpc'].mean()
highest_cpc = df_starter['cpc'].max()

print("Data Check for Section 2:")
print(f"Average Cost Per Click (CPC) across these pages: ${average_cpc:.2f}")
print(f"Highest single Cost Per Click in this dataset: ${highest_cpc:.2f}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

**What this model CAN do (Our Claims):**
* **Provide Decision Support:** It will rank our web pages so the editorial team knows exactly which articles to update first.
* **Identify Directional Trends:** It will spot which pages are showing patterns of fading traffic, decreasing impressions, or dropping ranks.
* **Surface Hidden Risks:** It will flag high-value pages (like those with high Cost Per Click) that are quietly losing steam before the traffic drop hurts company revenue.

**What this model NEVER will do (Our Boundaries):**
* **Claim Causal Proof:** It cannot prove *why* traffic dropped (e.g., it can't tell if a drop was caused by a Google algorithm change, a competitor's new backlink, or seasonal trends). It only flags *that* a drop is happening.
* **"Predict" Google:** We are not engineering a tool that reads Google's mind or forecasts future search engine updates. We are simply building an early-warning system based on historical performance data.
